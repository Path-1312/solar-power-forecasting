```python
import pandas as pd
import numpy as np

# Read the Excel file
df = pd.read_excel(
    r"/content/Forecasted table.xlsx"
)

# --------------------------------------------------
# Absolute Error
# Formula:
# |Actual - Predicted| / 90
# --------------------------------------------------

df['SARIMA_Absolute_Error'] = (
    np.abs(df['Actual'] - df['SARIMA']) / 90
)

df['LR_Absolute_Error'] = (
    np.abs(df['Actual'] - df['LR']) / 90
)

df['RF_Absolute_Error'] = (
    np.abs(df['Actual'] - df['RF']) / 90
)


# --------------------------------------------------
# 7% Tolerance
# --------------------------------------------------

tolerance = 0.07


df['SARIMA_Within_7_Percent'] = (
    df['SARIMA_Absolute_Error'] <= tolerance
)

df['LR_Within_7_Percent'] = (
    df['LR_Absolute_Error'] <= tolerance
)

df['RF_Within_7_Percent'] = (
    df['RF_Absolute_Error'] <= tolerance
)


# --------------------------------------------------
# Percentage of samples within 7% tolerance
# --------------------------------------------------

sarima_percentage = (
    df['SARIMA_Within_7_Percent'].mean() * 100
)

lr_percentage = (
    df['LR_Within_7_Percent'].mean() * 100
)

rf_percentage = (
    df['RF_Within_7_Percent'].mean() * 100
)


# --------------------------------------------------
# Number of samples within 7%
# --------------------------------------------------

sarima_count = df['SARIMA_Within_7_Percent'].sum()
lr_count = df['LR_Within_7_Percent'].sum()
rf_count = df['RF_Within_7_Percent'].sum()

total_samples = len(df)


# --------------------------------------------------
# Print Results
# --------------------------------------------------

print("========================================")
print("       7% TOLERANCE RESULTS")
print("========================================")

print(f"Total test samples: {total_samples}")
print()

print("SARIMA:")
print(f"Samples within 7%: {sarima_count}")
print(f"Percentage within 7%: {sarima_percentage:.2f}%")
print()

print("Linear Regression:")
print(f"Samples within 7%: {lr_count}")
print(f"Percentage within 7%: {lr_percentage:.2f}%")
print()

print("Random Forest:")
print(f"Samples within 7%: {rf_count}")
print(f"Percentage within 7%: {rf_percentage:.2f}%")
print()

```
